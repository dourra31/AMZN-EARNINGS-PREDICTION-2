```python
import numpy as np
from sklearn.linear_model import LogisticRegression

#X only includes RSI readings
X = np.array([
    [65],       # Apr 27, 2023
    [48],       # Aug 3, 2023
    [35],       # Oct 26, 2023
    [55],       # Feb 1, 2024
    [45],       # Apr 30, 2024
    [50],       # Aug 1, 2024
    [50],       # Oct 31, 2024
    [60.74]     # Feb 6, 2025
])

#1=stock went UP after earnings, 0 =DOWN
y = np.array([1, 1, 0, 0, 0, 0, 1, 0])


model = LogisticRegression()
model.fit(X, y)

#predict using only April 30 RSI
new_data = np.array([[50]])
prediction = model.predict(new_data)

if prediction[0] == 1:
    print("Model predicts: Amazon stock will go UP after earnings.")
else:
    print("Model predicts: Amazon stock will go DOWN after earnings.")
```
